# Graph Visualization with Different Colors and Styles

This notebook demonstrates various types of graphs and plots with different:
- Color palettes
- Line styles
- Markers
- Plot styles
- Themes

All graphs are generated with diverse styling to showcase visualization capabilities.

```python
# Import required libraries
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from matplotlib import cm
import warnings
warnings.filterwarnings('ignore')

# Set default figure size
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['figure.dpi'] = 100

print("Libraries imported successfully!")
```

```python
# Generate sample data for demonstrations
np.random.seed(42)

# Data for various plot types
categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
values1 = np.random.randint(50, 200, 5)
values2 = np.random.randint(30, 180, 5)
values3 = np.random.randint(40, 190, 5)

# Time series data
x = np.linspace(0, 10, 100)
y1 = np.sin(x) * 10 + 50
y2 = np.cos(x) * 10 + 50
y3 = np.sin(x + np.pi/4) * 10 + 50
y4 = np.cos(x + np.pi/4) * 10 + 50

# DataFrame for advanced plotting
df = pd.DataFrame({
    'Method': ['NLTK'] * 5 + ['SpaCy'] * 5 + ['TextBlob'] * 5,
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'Speed'] * 3,
    'Value': np.random.uniform(0.6, 0.95, 15)
})

print("Sample data generated successfully!")
print(f"\nDataFrame shape: {df.shape}")
print(df.head())
```

## 1. Bar Plots with Different Color Palettes

```python
# Bar plot with custom color palette - Viridis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# First bar plot with viridis colors
colors1 = plt.cm.viridis(np.linspace(0.2, 0.9, len(categories)))
bars1 = ax1.bar(categories, values1, color=colors1, edgecolor='black', linewidth=1.5)
ax1.set_title('Bar Plot with Viridis Color Palette', fontsize=14, fontweight='bold')
ax1.set_xlabel('Categories', fontsize=12)
ax1.set_ylabel('Values', fontsize=12)
ax1.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height)}', ha='center', va='bottom', fontsize=10)

# Second bar plot with custom color list
colors2 = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
bars2 = ax2.bar(categories, values2, color=colors2, edgecolor='navy', linewidth=1.5, alpha=0.8)
ax2.set_title('Bar Plot with Custom Hex Colors', fontsize=14, fontweight='bold')
ax2.set_xlabel('Categories', fontsize=12)
ax2.set_ylabel('Values', fontsize=12)
ax2.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels on bars
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height)}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Grouped bar plot with different colors
x_pos = np.arange(len(categories))
width = 0.25

fig, ax = plt.subplots(figsize=(14, 7))

bars1 = ax.bar(x_pos - width, values1, width, label='Series 1', 
               color='#e74c3c', edgecolor='black', linewidth=1.2)
bars2 = ax.bar(x_pos, values2, width, label='Series 2', 
               color='#3498db', edgecolor='black', linewidth=1.2)
bars3 = ax.bar(x_pos + width, values3, width, label='Series 3', 
               color='#2ecc71', edgecolor='black', linewidth=1.2)

ax.set_xlabel('Categories', fontsize=13, fontweight='bold')
ax.set_ylabel('Values', fontsize=13, fontweight='bold')
ax.set_title('Grouped Bar Plot with Three Series (Custom Colors)', fontsize=15, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels(categories)
ax.legend(loc='upper right', fontsize=11, framealpha=0.9)
ax.grid(axis='y', alpha=0.3, linestyle=':')

plt.tight_layout()
plt.show()
```

## 2. Line Plots with Different Styles and Markers

```python
# Line plots with different line styles and markers
fig, ax = plt.subplots(figsize=(14, 7))

# Different line styles and markers
ax.plot(x, y1, color='#e74c3c', linewidth=2.5, linestyle='-', 
        marker='o', markersize=6, markevery=10, label='Sine Wave (solid)', alpha=0.8)
ax.plot(x, y2, color='#3498db', linewidth=2.5, linestyle='--', 
        marker='s', markersize=6, markevery=10, label='Cosine Wave (dashed)', alpha=0.8)
ax.plot(x, y3, color='#2ecc71', linewidth=2.5, linestyle='-.', 
        marker='^', markersize=7, markevery=10, label='Shifted Sine (dash-dot)', alpha=0.8)
ax.plot(x, y4, color='#f39c12', linewidth=2.5, linestyle=':', 
        marker='D', markersize=6, markevery=10, label='Shifted Cosine (dotted)', alpha=0.8)

ax.set_xlabel('X Axis', fontsize=13, fontweight='bold')
ax.set_ylabel('Y Axis', fontsize=13, fontweight='bold')
ax.set_title('Line Plot with Various Line Styles and Markers', fontsize=15, fontweight='bold')
ax.legend(loc='upper right', fontsize=11, framealpha=0.9)
ax.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

## 3. Scatter Plots with Color Gradients

```python
# Scatter plot with color gradients
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Generate scatter data
n_points = 200
x_scatter = np.random.randn(n_points)
y_scatter = np.random.randn(n_points)
colors_scatter = np.random.rand(n_points)
sizes = np.random.randint(30, 200, n_points)

# First scatter plot with continuous colormap
scatter1 = ax1.scatter(x_scatter, y_scatter, c=colors_scatter, s=sizes, 
                       cmap='coolwarm', alpha=0.6, edgecolors='black', linewidth=0.5)
ax1.set_title('Scatter Plot with Coolwarm Colormap', fontsize=14, fontweight='bold')
ax1.set_xlabel('X Values', fontsize=12)
ax1.set_ylabel('Y Values', fontsize=12)
ax1.grid(True, alpha=0.3, linestyle='--')
plt.colorbar(scatter1, ax=ax1, label='Color Value')

# Second scatter plot with different colormap
scatter2 = ax2.scatter(x_scatter, y_scatter, c=colors_scatter, s=sizes, 
                       cmap='plasma', alpha=0.6, edgecolors='navy', linewidth=0.5)
ax2.set_title('Scatter Plot with Plasma Colormap', fontsize=14, fontweight='bold')
ax2.set_xlabel('X Values', fontsize=12)
ax2.set_ylabel('Y Values', fontsize=12)
ax2.grid(True, alpha=0.3, linestyle='--')
plt.colorbar(scatter2, ax=ax2, label='Color Value')

plt.tight_layout()
plt.show()
```

## 4. Seaborn Plots with Different Palettes

```python
# Seaborn bar plot with different palettes
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Palette 1: Set1
sns.barplot(data=df, x='Metric', y='Value', hue='Method', palette='Set1', ax=ax1)
ax1.set_title('Bar Plot with Set1 Palette', fontsize=13, fontweight='bold')
ax1.set_xlabel('Metrics', fontsize=11)
ax1.set_ylabel('Values', fontsize=11)
ax1.legend(title='NLP Method', fontsize=9)
ax1.grid(axis='y', alpha=0.3)

# Palette 2: Dark2
sns.barplot(data=df, x='Metric', y='Value', hue='Method', palette='Dark2', ax=ax2)
ax2.set_title('Bar Plot with Dark2 Palette', fontsize=13, fontweight='bold')
ax2.set_xlabel('Metrics', fontsize=11)
ax2.set_ylabel('Values', fontsize=11)
ax2.legend(title='NLP Method', fontsize=9)
ax2.grid(axis='y', alpha=0.3)

# Palette 3: husl
sns.barplot(data=df, x='Metric', y='Value', hue='Method', palette='husl', ax=ax3)
ax3.set_title('Bar Plot with HUSL Palette', fontsize=13, fontweight='bold')
ax3.set_xlabel('Metrics', fontsize=11)
ax3.set_ylabel('Values', fontsize=11)
ax3.legend(title='NLP Method', fontsize=9)
ax3.grid(axis='y', alpha=0.3)

# Palette 4: Custom color palette
custom_palette = ['#FF6B6B', '#4ECDC4', '#45B7D1']
sns.barplot(data=df, x='Metric', y='Value', hue='Method', palette=custom_palette, ax=ax4)
ax4.set_title('Bar Plot with Custom Palette', fontsize=13, fontweight='bold')
ax4.set_xlabel('Metrics', fontsize=11)
ax4.set_ylabel('Values', fontsize=11)
ax4.legend(title='NLP Method', fontsize=9)
ax4.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Box Plots and Violin Plots with Different Styles

```python
# Box plot and violin plot with custom colors
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Box plot
sns.boxplot(data=df, x='Method', y='Value', palette='Set2', ax=ax1, linewidth=2)
ax1.set_title('Box Plot with Set2 Palette', fontsize=14, fontweight='bold')
ax1.set_xlabel('NLP Method', fontsize=12)
ax1.set_ylabel('Performance Value', fontsize=12)
ax1.grid(axis='y', alpha=0.3, linestyle='--')

# Violin plot
sns.violinplot(data=df, x='Method', y='Value', palette='muted', ax=ax2)
ax2.set_title('Violin Plot with Muted Palette', fontsize=14, fontweight='bold')
ax2.set_xlabel('NLP Method', fontsize=12)
ax2.set_ylabel('Performance Value', fontsize=12)
ax2.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

## 6. Heatmaps with Different Color Schemes

```python
# Create correlation matrix-like data
np.random.seed(42)
data_matrix = np.random.rand(8, 8)
# Make it symmetric for correlation matrix appearance
data_matrix = (data_matrix + data_matrix.T) / 2
np.fill_diagonal(data_matrix, 1)

labels = ['Token', 'Lemma', 'Stem', 'POS', 'NER', 'Dep', 'Vector', 'Sentiment']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Heatmap 1 with coolwarm
im1 = ax1.imshow(data_matrix, cmap='coolwarm', aspect='auto', interpolation='nearest')
ax1.set_xticks(np.arange(len(labels)))
ax1.set_yticks(np.arange(len(labels)))
ax1.set_xticklabels(labels, rotation=45, ha='right')
ax1.set_yticklabels(labels)
ax1.set_title('Heatmap with Coolwarm Colormap', fontsize=14, fontweight='bold')
plt.colorbar(im1, ax=ax1, label='Correlation Value')

# Heatmap 2 with RdYlGn
im2 = ax2.imshow(data_matrix, cmap='RdYlGn', aspect='auto', interpolation='nearest')
ax2.set_xticks(np.arange(len(labels)))
ax2.set_yticks(np.arange(len(labels)))
ax2.set_xticklabels(labels, rotation=45, ha='right')
ax2.set_yticklabels(labels)
ax2.set_title('Heatmap with RdYlGn Colormap', fontsize=14, fontweight='bold')
plt.colorbar(im2, ax=ax2, label='Correlation Value')

plt.tight_layout()
plt.show()
```

```python
# Seaborn heatmap with annotations
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Heatmap with viridis
sns.heatmap(data_matrix, annot=True, fmt='.2f', cmap='viridis', 
            xticklabels=labels, yticklabels=labels, ax=ax1, 
            cbar_kws={'label': 'Value'}, linewidths=0.5)
ax1.set_title('Annotated Heatmap with Viridis Colormap', fontsize=13, fontweight='bold')

# Heatmap with Blues
sns.heatmap(data_matrix, annot=True, fmt='.2f', cmap='Blues', 
            xticklabels=labels, yticklabels=labels, ax=ax2, 
            cbar_kws={'label': 'Value'}, linewidths=0.5)
ax2.set_title('Annotated Heatmap with Blues Colormap', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 7. Area and Stack Plots with Custom Colors

```python
# Area plots with different colors
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Simple area plot
ax1.fill_between(x, y1, color='#e74c3c', alpha=0.5, label='Series 1')
ax1.fill_between(x, y2, color='#3498db', alpha=0.5, label='Series 2')
ax1.set_title('Area Plot with Transparent Fills', fontsize=14, fontweight='bold')
ax1.set_xlabel('X Axis', fontsize=12)
ax1.set_ylabel('Y Axis', fontsize=12)
ax1.legend(loc='upper right', fontsize=11)
ax1.grid(True, alpha=0.3, linestyle='--')

# Stacked area plot
y_stack1 = np.abs(np.sin(x)) * 20
y_stack2 = np.abs(np.cos(x)) * 15
y_stack3 = np.abs(np.sin(2*x)) * 10

ax2.fill_between(x, 0, y_stack1, color='#FF6B6B', alpha=0.7, label='Layer 1')
ax2.fill_between(x, y_stack1, y_stack1 + y_stack2, color='#4ECDC4', alpha=0.7, label='Layer 2')
ax2.fill_between(x, y_stack1 + y_stack2, y_stack1 + y_stack2 + y_stack3, 
                 color='#45B7D1', alpha=0.7, label='Layer 3')
ax2.set_title('Stacked Area Plot with Custom Colors', fontsize=14, fontweight='bold')
ax2.set_xlabel('X Axis', fontsize=12)
ax2.set_ylabel('Y Axis', fontsize=12)
ax2.legend(loc='upper right', fontsize=11)
ax2.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

## 8. Pie Charts with Different Color Schemes

```python
# Pie charts with custom colors and styles
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Data for pie charts
sizes = [30, 25, 20, 15, 10]
labels_pie = ['Tokenization', 'Lemmatization', 'POS Tagging', 'NER', 'Parsing']

# Pie chart 1 with custom colors
colors_pie1 = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
explode = (0.1, 0, 0, 0, 0)  # explode 1st slice

wedges1, texts1, autotexts1 = ax1.pie(sizes, explode=explode, labels=labels_pie, 
                                       colors=colors_pie1, autopct='%1.1f%%',
                                       shadow=True, startangle=90)
ax1.set_title('Pie Chart with Custom Colors and Shadow', fontsize=14, fontweight='bold')

# Make percentage text bold
for autotext in autotexts1:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(11)

# Pie chart 2 with different style
colors_pie2 = plt.cm.Set3(np.linspace(0, 1, len(labels_pie)))
wedges2, texts2, autotexts2 = ax2.pie(sizes, labels=labels_pie, 
                                       colors=colors_pie2, autopct='%1.1f%%',
                                       startangle=45, 
                                       wedgeprops={'edgecolor': 'black', 'linewidth': 2})
ax2.set_title('Pie Chart with Set3 Palette and Borders', fontsize=14, fontweight='bold')

for autotext in autotexts2:
    autotext.set_color('black')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(11)

plt.tight_layout()
plt.show()
```

## 9. Histograms with Different Colors and Styles

```python
# Histograms with different styles
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Generate sample data
data1 = np.random.normal(100, 15, 1000)
data2 = np.random.normal(110, 20, 1000)
data3 = np.random.normal(105, 18, 1000)

# Histogram 1: Simple with custom color
ax1.hist(data1, bins=30, color='#e74c3c', alpha=0.7, edgecolor='black')
ax1.set_title('Histogram with Red Color', fontsize=13, fontweight='bold')
ax1.set_xlabel('Value', fontsize=11)
ax1.set_ylabel('Frequency', fontsize=11)
ax1.grid(axis='y', alpha=0.3, linestyle='--')

# Histogram 2: Multiple datasets
ax2.hist([data1, data2, data3], bins=30, color=['#e74c3c', '#3498db', '#2ecc71'], 
         alpha=0.6, label=['Dataset 1', 'Dataset 2', 'Dataset 3'], edgecolor='black')
ax2.set_title('Overlapping Histograms with Multiple Colors', fontsize=13, fontweight='bold')
ax2.set_xlabel('Value', fontsize=11)
ax2.set_ylabel('Frequency', fontsize=11)
ax2.legend(fontsize=10)
ax2.grid(axis='y', alpha=0.3, linestyle='--')

# Histogram 3: Step style
ax3.hist(data1, bins=30, histtype='step', color='#9b59b6', linewidth=2.5)
ax3.hist(data2, bins=30, histtype='step', color='#e67e22', linewidth=2.5)
ax3.set_title('Step-style Histogram', fontsize=13, fontweight='bold')
ax3.set_xlabel('Value', fontsize=11)
ax3.set_ylabel('Frequency', fontsize=11)
ax3.grid(axis='y', alpha=0.3, linestyle='--')

# Histogram 4: Stacked
ax4.hist([data1, data2], bins=30, stacked=True, 
         color=['#1abc9c', '#f39c12'], alpha=0.7, 
         label=['Dataset 1', 'Dataset 2'], edgecolor='black')
ax4.set_title('Stacked Histogram', fontsize=13, fontweight='bold')
ax4.set_xlabel('Value', fontsize=11)
ax4.set_ylabel('Frequency', fontsize=11)
ax4.legend(fontsize=10)
ax4.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

## Summary

This notebook demonstrated various graph types with different:

### Color Schemes Used:
- **Matplotlib colormaps**: viridis, coolwarm, plasma, RdYlGn, Blues, Set1, Set2, Set3, Dark2, HUSL, muted
- **Custom hex colors**: #FF6B6B, #4ECDC4, #45B7D1, #FFA07A, #98D8C8, #e74c3c, #3498db, #2ecc71, etc.

### Styles and Features:
- **Line styles**: solid (-), dashed (--), dash-dot (-.), dotted (:)
- **Markers**: circles (o), squares (s), triangles (^), diamonds (D)
- **Transparency**: alpha values from 0.5 to 0.8
- **Edge colors**: black, navy, white
- **Grid styles**: dashed, dotted, with varying alpha
- **Shadows**: on pie charts
- **Annotations**: on heatmaps

### Graph Types Covered:
1. Bar plots (single, grouped, with various palettes)
2. Line plots (with different styles and markers)
3. Scatter plots (with color gradients)
4. Box plots and violin plots
5. Heatmaps (multiple colormaps)
6. Area and stacked area plots
7. Pie charts (with explosions and borders)
8. Histograms (overlapping, step-style, stacked)

All graphs are designed to be visually distinct with different color combinations and styling options!
